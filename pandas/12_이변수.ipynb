# 이변수 데이터 분석

```python
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
import platform

if platform.system() == 'Windows':
    matplotlib.rc('font', family='Malgun Gothic')
elif platform.system() == 'Darwin':
    matplotlib.rc('font', family='AppleGothic')
else:
    matplotlib.rc('font', family='DejaVu Sans')

plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 100

print('NumPy 버전:', np.__version__)
print('환경 설정 완료')
```

- kobis 데이터 로드

```python
kobis_df = pd.read_csv('data/kobis.csv')
kobis_df.head()
kobis_df.shape
kobis_df.info()
```

```python
kobis_df['일일매출액'].value_counts()
kobis_df['일일매출액'].plot.box()
plt.show()
```

```python
kobis_df['일일관객수'].value_counts()
```

```python
# 일일매출액, 일일관객수의 평균값, 중앙값
kobis_df[['일일매출액','일일관객수']].mean()
kobis_df[['일일매출액','일일관객수']].median()
kobis_df[['일일매출액','일일관객수']].agg(['mean', 'median'])
```

```python
kobis_df.describe()
```

```python
# 일일매출액, 일일관객수의 상관관계에 따른 산점도
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

kobis_df.plot.scatter(
    ax = axes[0],
    x='일일매출액',
    y='일일관객수',
    title='일일매출액 vs 일일관객수 - 산점도'
)
kobis_df.plot.scatter(
    ax = axes[1],
    x='일일매출액',
    y='일일관객수',
    marker='s',
    s=50,
    c='coral',
    title='일일매출액 vs 일일관객수 - 산점도'
)
plt.show()
```

```python
# 일일매출액과 일일관객수의 상관계수
r = kobis_df['일일매출액'].corr(kobis_df['일일관객수'])
print(f"r(상관계수): {r:.3f}")  ## 이상치 데이터에 영향을 받음
```

#### 기온에 따른 따릉이 대여건수 산점도, 상관계수, 회귀식 도출
- 따릉이 데이터 + 날씨 데이터

```python
# 데이터셋 로드
df1 = pd.read_csv('data/서울특별시 공공자전거 일별 대여건수_25.1-6.csv')
df2 = pd.read_csv('data/OBS_ASOS_DD_20250101.csv')
```

```python
df1.head()
```

```python
df2.head()
```

```python
print(df1.info(), '\n', df2.info())
```

```python
# df1, df2 데이터 타입 변환 - to_datetime()
df1['대여일자'] = pd.to_datetime(df1['대여일자'])
```

```python
df2['일시'] = pd.to_datetime(df2['일시'])
```

```python
print(df1['대여일자'].dtypes)
print(df2['일시'].dtypes)
```

```python
# df1, df2 컬럼명 변경 - rename()
df1.rename(columns = {'대여일자':'날짜'}, inplace=True)
df2.rename(columns = {'일시':'날짜', '평균기온(°C)':'기온'}, inplace=True)
print(df1, df2)
```

```python
df = pd.merge(df1, df2, how='inner', on='날짜')
```

```python
df[['기온', '대여건수']].describe()
```

```python
df['대여건수'].plot.box()
plt.show()
```

```python
r = df['기온'].corr(df['대여건수'])
print(f"r : {r:.3f}")
```

```python
# seaborn 패키지 사용
import seaborn as sns

sns.scatterplot(
    x='기온',
    y='대여건수',
    data=df
)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12,5))

sns.scatterplot(
    x='기온',
    y='대여건수',
    data = df,
    ax = axes[0],
)

sns.regplot(
    x='기온',
    y='대여건수',
    data = df,
    ax = axes[1],
)

# sns.lmplot(
#     x='기온',
#     y='대여건수',
#     data = df,
#     ax = axes[2],
#     hue=''
# )
plt.show()
```

```python
# y = mx + b
# m기울기, b절편 을 구하는 식
# numpy.polyfit(x, y)

m, b = np.polyfit(df['기온'], df['대여건수'], 1)
print(m, b)
```

```python
# 기온이 25일 떄 대여건수 얼마나 될까?
# 회귀식 : y(대여건수) = mx(기온) + b
x = 25
y = m*x+b 
print(f"기온이 25일 때 예측된 대여건수 : {y:.0f}")
```
