# 결손 데이터(결측치) 처리

### 결손데이터(결측치)란?
- 결측치(Missing Data)는 값이 존재하지 않는 데이터를 의미
- 판다스에는 NaN으로 표현됨

```python
import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd
import numpy as np
```

## 1. 결손 데이터 생성

```python
df = pd.DataFrame(
    {
        'Name':['Smith','Lee','Kim', 'Park'],
        'Age':[20, np.nan, 30, 40],
        'Score':[90, 85, np.nan, 75],
    }
)
df
```

```python
df.info()
```

```python
df.describe()
```

### 2. 결손 데이터 확인
- isna() = isnull()

```python
df
df.isna()
```

```python
df.isnull()
```

```python
# NaN 값이 있는 데이터만 합계로 계산
print(df.isna().sum())
print(df.isnull().sum())
```

### 3. 결손 데이터 삭제
- dropna() : 결손 데이터가 포함된 행 또는 열 삭제

```python
df
```

```python
filter_na = df.dropna()
```

```python
print(df)
print('-'*30)
print(filter_na)
```

```python
# 결측치 데이터가 포함된 열 삭제
filter_na_cols = df.dropna(axis=1)
print(df)
print('-'*30)
print(filter_na_cols)
```

```python
# 결측치가 저장된 특정 컬럼만 삭제
df.dropna(axis=0, subset=['Age'])
```

```python
# Addr 컬럼 추가, 값은 모두 결측치(NaN)
df['Addr'] = np.nan
df
```

```python
# 모든 데이터가 결측치인 컬럼 삭제
df.dropna(how='all', axis=1, inplace=True)
```

```python
# 하나라도 결측치를 포함한 경우 삭제 how = 'any', axis = 0, inplace=False
df.dropna(how='any', axis=0, inplace=False)
```

### 4. 결손 데이터 대체
- fillna() : 결손 데이터를 특정값으로 대체

```python
# 결손 데이터를 0으로 대체
df.fillna(0)
df
```

```python
# 결손 데이터를 'Unknown'으로 모두 대체
df_str = df.fillna('Unknown')
df_str.info()
```

```python
# Age 컬럼의 결손 데이터만 0으로 대체
df['Age'] = df['Age'].fillna(0)
```
