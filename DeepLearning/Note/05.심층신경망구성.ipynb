#### 패션 MNIST

```python
from tensorflow import keras

(train_input, train_target), (test_input, test_target) = \
    keras.datasets.fashion_mnist.load_data()
```

```python
# train과 valid data 분리
from sklearn.model_selection import train_test_split

train_scaled = (train_input / 255.0).reshape(-1, 28*28)
train_scaled, val_scaled, train_target, val_target = \
    train_test_split(
        train_scaled,
        train_target,
        test_size=0.2,
        random_state=42
    )
```

#### Layer를 추가하는 방법 -1

```python
# Layer 만들기
dense1 = keras.layers.Dense(100, activation='sigmoid', input_shape=(784,)) # 은닉층
dense2 = keras.layers.Dense(10, activation='softmax') # 출력층
```

```python
# 전체 모들에 층울 추가
model = keras.Sequential([dense1, dense2])
```

```python
# Summary
model.summary()
```

#### Layer를 추가하는 방법 - 3

```python
model = keras.Sequential()
model.add(keras.layers.Dense(100, activation='sigmoid', input_shape=(784,), name='hidden'))
model.add(keras.layers.Dense(10, activation='softmax', name='output'))
```

```python
model.summary()
```

> 제일 많이 사용하는 방법이다. 파이썬 코드가 중간에 삽입이 가능하다.

```python
model.compile(
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
model.fit(
    train_scaled,
    train_target,
    epochs=5
)
```

#### 입력층을 이미지 그대로 사용하기

```python
(train_input, train_target), (test_input, test_target) = \
    keras.datasets.fashion_mnist.load_data()

train_scaled = (train_input / 255.0)
train_scaled, val_scaled, train_target, val_target = \
    train_test_split(
        train_scaled,
        train_target,
        test_size=0.2,
        random_state=42
    )
```

```python
train_scaled.shape
```

```python
# epochs 수정하고 다시 시작하는 부분
model = keras.Sequential()
model.add(keras.layers.Flatten(input_shape=(28, 28)))
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dense(50, activation='relu'))
model.add(keras.layers.Dense(30, activation='relu'))
model.add(keras.layers.Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
# 손실함수
model.compile(
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
# 학습하기
model.fit(
    train_scaled,
    train_target,
    epochs=7
)
```

```python
# 검증데이터로 확인
score_train = model.evaluate(train_scaled, train_target)
print("loss =", score_train[0])
print("accuracy =", score_train[1])
# epochs 7 -> acc : 0.942
# epochs 10 -> acc : 0.936
# epochs 15 -> acc : 0.918
# epochs 20 -> acc : 0.929
```

```python
# 테스트로 확인
score_test = model.evaluate(test_input, test_target)
print("loss =", score_test[0])
print("accuracy =", score_test[1])
# epochs 7 -> acc : 0.858
# epochs 10 -> acc : 0.859
# epochs 15 -> acc : 0.851
# epochs 20 -> acc : 0.858
```

```python
# valid 확인
score_val = model.evaluate(val_scaled, val_target)
print("loss =", score_val[0])
print("accuracy =", score_val[1])
# epochs 7 -> acc : 0.879
# epochs 10 -> acc : 0.881
# epochs 15 -> acc : 0.880
# epochs 20 -> acc : 0.883
```
